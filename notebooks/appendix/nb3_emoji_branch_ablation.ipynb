```python
# C5b: Emoji2Vec Coverage Statistics
# Assume train/val/test are loaded from CSV
DATA_DIR = Path("../../data/vigoemotions")

dfs = {
    "Train": pd.read_csv(DATA_DIR / "train.csv"),
    "Val": pd.read_csv(DATA_DIR / "val.csv"),
    "Test": pd.read_csv(DATA_DIR / "test.csv")
}
dfs["Full"] = pd.concat(dfs.values())

# Mock emoji2vec vocab for demonstration (in reality, load from bin)
emoji2vec_vocab_set = set(["😀", "😂", "😢", "😭", "😍"]) 

results = []
for name, df in dfs.items():
    stats = emoji_coverage(df['text'].tolist(), emoji2vec_vocab_set)
    results.append({
        "Split": name,
        "Total Occurrences": stats["total_occurrences"],
        "Unique Types": stats["unique_types"],
        "Coverage (Occ) %": f"{stats['coverage_by_occurrence']:.2f}%",
        "Coverage (Type) %": f"{stats['coverage_by_type']:.2f}%",
        "Sentences with OOV %": f"{stats['pct_sentences_with_oov']:.2f}%"
    })

df_res = pd.DataFrame(results)
print(to_latex_table(df_res, "Emoji2Vec Coverage Statistics.", "tab:emoji_coverage"))
```
